# Test Protocol for PBMC yeild comparison

## Best prepration for Bulk assay
+ First make a stock of mixed ChronoSeq V5 beads 200&mu;l at 450 beads/&mu;l in Lysis Buffer
+ Take 40&mu;l and split across four replicates 10&mu;l each. This will be our control condition 4 replicates.
+ Lets call these tubes C1 through C4.
+ Next we need to make 100&mu;l of beads with 3U/&mu;l of RNAase inhibitor. For this we will use Watchmaker RNAase inhibitor we have left over at 800U/&mu;l. We will also need 5mM DTT. Since this needs to be added fresh we will prepare this final preparation **after preparing the PBMC in two different conditions below**. 
+ To make 3U/&mu;l we need 300U for 100&mu;l. So we will add:
    + 0.375&mu;l of RNAase inhibitor using the P2.5
    + 0.5&mu;l of 1M DTT from frozen stock for final concentration of 5mM
+ Next we need to split this volume into 8 Tubes with 10&mu;l of beads each.
+ Inhibitor only will be R1 through R4
+ Inhibitor plus DPBS only for PBMC tubes will be D1 through D4. 

## Preparation of PBMC for the two conditions
+ Make sure cell reservoirs are clean before starting.
+ Also make sure you have enough RT buffer for washing 12 Tubes. Need 24ml Volume at least for 2X RT buffer. If you don't have enough make more. 
+ Order and receive fresh PBMC from ATCC the next day. Ideally order on Monday afternoon or after 6pm so they ship it on Tuesday or Wednesday and we get the cells on Wednesday or Thursday.

### Cell Suspension Preparation
+ Check the reagents to make sure you have enough. Also make sure the Ficoll is not contaminated. 
+ Prep the ChronoSeq device make sure 5% CO<sub>2</sub> is connected and device and water bath is turned on.
+ Get some ice and put two tubes of LPS on ice. We will be stimulating with 1&mu;g/ml of LPS for this experiment.
+ You will need warmed DPBS and DMEM with 1% PEN-STREP.
+ We need to start working quickly with the freshly arrived PBMC from ATCC.
+ Wash the PBMC Vial with DPBS to get any remaining cells out.
+ Resuspend the cells in DPBS then split the cells into two equal parts.
+ For the first part wash the cells twice with DMEM 1% PEN-STREP 
+ For the second part wash the cells twice with DPBS 1% PEN-STREP
+ Make teh final volume 1000cells/&mu;l for both tubes.
+ I expect this to be close to 12ml for each Tube.
+ Put two clean magnets into two different 50ml Tubes and transfer the cells to each tubes for DMEM or DPBS.
+ Load the cells into the ChronoSeq device with 5%CO<sub>2</sub> pressure source connected. 
+ Stimulate the cells with LPS. We will wait 1 hour before taking samples. 
+ In the meantime you can prepare your beads and add DTT and Watchmaker RNAase inhibitor.
+ Following the same Bulk RNA-Seq protocol wash once with 6X SSC and then twice with 2X RT buffer.
+ We will use 1/4 the reagents we normally use for an experiment Single-cell split. Just use a quarter of each as the Single-Cell protocol. This mean final volume should be 25&mu;l and then add all reagents as single-cell. Don't pool the different conditions into the same reaction. 
+ Run the RT reaction and get the qPCR yeilds.

### Instruments used: 
+ Bio-Rad CFX Opus 96 

### Cell Culture and Cell Preparation
+ We will need to culture THP-1 cells and K562 Cells for this experiment. 
+ Make sure you have enough K562 Cells on the day of the experiment to make at least 7ml of K562 at 430cells/&mu;l
+ We will follow this [protocol for culturing the THP-1 cells](protocol_for_preparing_cells_AdditionalFiltration.ipynb#Protocol-for-Preparing-THP-1-cells-after-PMA-differenciation-with-LPS-Stimulation).
    + However, keep the THP-1 and K562 cell suspensions separated for this experiment. We want qPCR to only measure the effect on K562 cells.
+ The goal of this experiment is to see if K562 cells can be stimulated by THP-1 Cells through Paracrine signalling.

### On Day of Experiment 
+ Make 6 K562 1.5ml Tubes with DMEM 1%PEN-STREP with 0.5ml of K562 Cells at 430cells/&mu;l in each:
    + We will label these tubes 3hr, 2hr, 1hr, 0hr, No Stimulation and TNF&alpha;
    + Keep these tubes with lid open inside an incubator at 37&deg;C and 5%CO<sub>2</sub>
+ Make enough Lysis reagent for 13 samples and keep on ice.
+ Thaw LPS and TNF&alpha; Vials and then keep on ice.
+ Load the ChronoSeq device reservoir 1 with 40ml of THP-1 Cell suspension at 430cells/&mu;l similar to [Bulk ChronoSeq](protocol_and_software_for_running_chronoseq_device.ipynb#bulk_cell) however, we will be taking samples manually.
    
### Cell Lysis
+ We will follow a protocol [similar to SingleShot Cell Lysis Kit](https://www.bio-rad.com/sites/default/files/webroot/web/pdf/lsr/literature/10000112808.pdf) but we will not change our media to DPBS.
+ This is what our overall protocol looks like:
    + THP-1 cell suspension stimulated with 1&mu;g/ml LPS.
    + Samples of THP-1 Cells taken at 0hr, 1hr, 2hr and 3hr.
    + Supernatant of THP-1 samples then added to K562 Cell suspension.
    + K562 NFKBIA and ACTB Genes measured at 0hr after adding THP-1 Supernatant, and 1hr after adding THP-1 Supernatant.
+ Prepare 12 PCR Tubes for Cell Lysis as follows and keep these Tubes on ice while your prepare. For each tube:
    + Label the Tubes 3H_0,3H_1,2H_0,2H_1,1H_0,1H_1,0H_0,0H_1,NS_0,NS_1,TNF_0,TNF_1.
    + 48&mu;l Single Shot Cell Lysis Buffer
    + 1&mu;l Proteinase K Solution
    + 1&mu;l DNase Solution
+ For each tube we will:
    + Add 5&mu;l of Cell Suspension. Mix Well. This is equivalent to 1075 K562 Cells.
    + Keep the tube back on ice till you have collected all 12 samples.
    
+ We need to take both 0hr and 1hr samples for K562 as well. This will help calculate the impact of LPS stimulation alone as well. We will only take 0.9ml Samples of THP-1 cells to conserve the amount left for our experiment. Always keep the K562 tubes back at 37&deg;C and 5%CO<sub>2</sub>. 

|Time|THP-1 Sample|K562 5&mu;l Sample for qPCR|
|-|-|-|
|0hr|Add LPS take 0.9 ml sample spin down at 1500g for 1 minute and then add 500&mu;l Supernatant to K562 0hr Epie|Immediately also take sample for K562 0H_0|

|1hr|take 0.9 ml sample spin down at 1500g for 1 minute and then add 500&mu;l Supernatant to K562 1hr flask|Immediately take sample from 1hr tube for 1H_0 and then take sample from K562 0hr tube for 0H_1|
|1hr30mins|Do nothing|Take samples from No Stimulation NS_1 and TNF&alpha K562 Tube for TNF_1|
|2hr|take 0.9 ml sample spin down at 1500g for 1 minute and then add 500&mu;l Supernatant to K562 2hr flask|Immediately take sample from 2hr tube for 2H_0 and then take sample from K562 1hr tube for 1H_1|
|3hr|take 0.9 ml sample spin down at 1500g for 1 minute and then add 500&mu;l Supernatant to K562 3hr flask|Immediately take sample from 3hr tube for 3H_0 and then take sample from K562 2hr tube for 2H_1|
|4hr|Do nothing|Take sample from K562 3hr tube for 3H_1|

+ Spin down all the Tubes.
    + Using a PCR Machine run the following program:
        + Lid Temperature 105&deg;C
        + Volume 55&mu;l
        + Incubate for 10 min at 21&deg;C
        + Followed by 5 min at 37°C,
        + 5 min at 75°C.
        + 4&deg;C for &infin;
    + The cell lysate can be stored on ice for up to 4 hr, at -20°C for up to 2 months, or at -80°C for up to 12 months.
    + After PCR Immediately store the Tubes at -80&deg;C

### Single Step RT-qPCR
+ Adapted [from existing protocol here](https://www.bio-rad.com/webroot/web/pdf/lsr/literature/10032048.pdf).
+ The iTaq kit comes with ROX as a background reference dye so remember to set that in your qPCR machine. ROX is not necessary for Bio-Rad CFX Opus 96. 
+ For each conditon you will have to setup 4 qPCR reactions. Two replicates for each gene NFKBIA and ACTB. For each 20&mu;l reaction PCR Tube:
    + iTaq™ Universal SYBR® Green reaction mix 2X: 10&mu;l
    + 1&mu;l PrimePCR Assay primers (20X).
    + iScript reverse transcriptase 0.25&mu;l
    + 4&mu;l Cell Lysate.
    + 4.75&mu;l Nuclease Free Water
+ Setup the qPCR reaction as follows: 
    + 10min at 50&deg;C for Reverse Transcription.
    + 1min at 95&deg;C for DNA denaturation and Polymerse activation.
    + 40 Cycles of:
        + 95&deg;C 10seconds Denaturation
        + 60&deg;C 30seconds Annealing and Plate Read
+ Here is a screenshot of the plate setup. <br><img src="qPCR_validation_files/K562_Costimulation_2_Samples/Capture.PNG">
+ After the plate setup. The plate was sealed with a Optically clear adhesive film.
    + Make sure the lips are completely sealed to avoid leaks. Then vortex the plate.
+ Plate was then spun down and loaded into the qPCR machine.
+ Run the Program.
